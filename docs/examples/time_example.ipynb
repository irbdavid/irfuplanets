```python
%matplotlib widget
import matplotlib.pyplot as plt
import numpy as np
import irfuplanets.time as time
```

```python
# Convert a UTC string to spiceet
t = time.spiceet("2014-03-12T00:30")
print(t)

# As a datetime64
print(time.spiceet_to_datetime64(t))

# Or using the central time_convert function 
print(time.time_convert(t, output_format="datetime64"))
print(time.time_convert(t, output_format="datetime"))

# Helper object
c = time.CelsiusTime(t)
print(f"{c.year=}, {c.day=}, {c.month=}, {c.doy=}, {c.spiceet=}")

# Back to utcstr:
print(time.utcstr(t))
print(time.utcstr(t, fmt="ISOC"))
```

```python
t = np.arange(
    time.spiceet("2015-01-01T00:30"),
    time.spiceet(np.datetime64("2015-01-02T00:00")),
    600.
)
print(t.shape)
y = np.sin(t * 2. * np.pi / 3600.)
plt.figure()
plt.plot(t, y)
plt.xlim(t[0], t[-1])
time.setup_time_axis(calendar=True, verbose=True)
plt.show()
```
